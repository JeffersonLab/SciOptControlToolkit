```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
path = "../results/test_MOBO/trial_0/"
```

```python
for i in range(900, 1000):
    results = np.load(os.path.join(path, "inference_results_"+str(i).zfill(6)+".pkl"), allow_pickle=True)
    heat = results['heat']
    trip = results['trip']
    if i == 0:
        plt.scatter(heat[:128], trip[:128], label="initial", c=results['energy'][:128])    
        plt.colorbar()
        plt.title("Initial points", fontsize=18)
        plt.xlabel("Heat Load [W]", fontsize=14)
        plt.ylabel("Trip Rate [per hour]", fontsize=14)
        plt.legend(fontsize=14)
        plt.show()
    plt.scatter(heat[:128], trip[:128], color="gray", label="initial")
    plt.scatter(heat[128:], trip[128:], label="from MOBO", c=results['energy'][128:])
    plt.colorbar()
    plt.title("Step "+str(i+1), fontsize=18)
    plt.xlabel("Heat Load [W]", fontsize=14)
    plt.ylabel("Trip Rate [per hour]", fontsize=14)
    plt.legend(fontsize=14)
    plt.show()
    
```
